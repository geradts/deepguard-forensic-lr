## Official DF40 processed testing data

This is the folder supplied by the DF40 authors: ~93 GB, all 40 methods, fake data after frame extraction and face cropping.


```python
DF40_FOLDER='https://drive.google.com/drive/folders/1U8meBbqVvmUkc5GD0jxct6xe6Gwk9wKD?usp=drive_link'
print('Official source:', DF40_FOLDER)
print('Downloading directly into:', DEST)
```

```python
# gdown writes directly to the mounted Google Drive destination.
# --continue allows individual interrupted files to resume when supported.
import subprocess, sys
cmd=[sys.executable,'-m','gdown','--folder',DF40_FOLDER,'-O',str(DEST),'--remaining-ok']
print(' '.join(cmd))
result=subprocess.run(cmd)
if result.returncode!=0:
    raise RuntimeError(f'DF40 download failed with exit code {result.returncode}')
```

```python
files=[p for p in DEST.rglob('*') if p.is_file()]
size=sum(p.stat().st_size for p in files)
print('Files:',len(files))
print('Size:',round(size/1024**3,2),'GiB')
print('Top-level entries:')
for p in list(DEST.iterdir())[:30]: print(' ',p.name)
if size < 80*1024**3:
    print('WARNING: less than 80 GiB detected; the official test set is approximately 93 GB. Check whether the Drive folder download was complete.')
else:
    print('Dataset size check: plausible for the official processed test set.')
```
